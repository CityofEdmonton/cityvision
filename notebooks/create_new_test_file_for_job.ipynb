```python
import os
import cv2
```

```python
folder_name = "../916_SCU1F4_2025-05-20_2345"
target_folder = "../916_SCU1F4_2025-05-20_2345_test_10"
```

```python
# read all video files in the folder and cut to 1 minute
video_files = [f for f in os.listdir(folder_name) if f.endswith('.mp4')]
for video_file in video_files:
	full_path = os.path.join(folder_name, video_file)
	# Here you would add the code to cut the video to 1 minute
	print(f"Processing {full_path}...")  # Placeholder for actual processing logic
	# use cv2 to read the video and cut it to 1 minute
	# create a new folder for the test files
	if not os.path.exists(target_folder):
		os.makedirs(target_folder)
	# read the video frame by frame
	cap = cv2.VideoCapture(full_path)
	if not cap.isOpened():
		print(f"Error opening video file {full_path}")
		continue
	frame_rate = cap.get(cv2.CAP_PROP_FPS)
	total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
	one_minute_frames = int(frame_rate * 10)  # Calculate frames for 10 seconds
	output_file = os.path.join(target_folder, video_file)
	fourcc = cv2.VideoWriter_fourcc(*'mp4v')
	out = cv2.VideoWriter(output_file, fourcc, frame_rate, (int(cap.get(cv2.CAP_PROP_FRAME_WIDTH)),
															int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))))
	frame_count = 0
	while cap.isOpened() and frame_count < one_minute_frames:
		ret, frame = cap.read()
		if not ret:
			break
		out.write(frame)
		frame_count += 1
	cap.release()
	out.release()
	print(f"Saved cut video to {output_file}")	
```
